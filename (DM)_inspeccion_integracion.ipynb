# <font color=blue> INSPECCIÓN INICIAL E INTEGRACIÓN DE FUENTES</font>

### <font color=green> 1. Importar datos </font>

```python
# Librería principal para trabajar con datos en tablas
import pandas as pd

# Para que Pandas muestre más columnas en pantalla
#pd.set_option("display.max_columns", 100)

# ============================================================
#  LEER LAS FUENTES DE DATOS
# ============================================================

df1 = pd.read_csv("fuente1_estudiantes.csv")
df2 = pd.read_csv("fuente2_academica.csv")

print("Fuente 1:")
display(df1.head())

print("Fuente 2:")
df2.head()
```

### <font color=green> 2. REVISAR EL TAMAÑO DE CADA FUENTE </font>

```python
# shape devuelve: (número de filas, número de columnas)

print("Tamaño de fuente 1:", df1.shape)
print("Tamaño de fuente 2:", df2.shape)
```

### <font color=green> 3. REVISAR LOS TIPOS DE DATOS DETECTADOS </font>
######Importante: Que una variable aparezca como número no significa siempre que sea una variable numérica.
###### Ejemplo: `id_estudiante es un código o identificador.`


```python
print("Tipos de datos en fuente 1:")
display(df1.dtypes)

print("Tipos de datos en fuente 2:")
df2.dtypes
```

### <font color=green> 4. USAR describe() PARA RESUMIR CADA FUENTE </font>

```python
# describe(include="all") resume variables numéricas y categóricas.

print("Resumen de fuente 1:")
display(df1.describe(include="all"))

print("Resumen de fuente 2:")
display(df2.describe(include="all"))
```

### <font color=green> 5. REVISAR VALORES FALTANTES </font>

```python
print("Valores faltantes en fuente 1:")
display(df1.isnull().sum())

print("Valores faltantes en fuente 2:")
display(df2.isnull().sum())


# También podemos ver el porcentaje de valores faltantes.
print("Porcentaje de valores faltantes en fuente 1:")
display((df1.isnull().mean() * 100).round(2))

print("Porcentaje de valores faltantes en fuente 2:")
display((df2.isnull().mean() * 100).round(2))
```

### <font color=green> 6. REVISAR CANTIDAD DE VALORES DISTINTOS POR VARIABLE </font>

```python
# Esto ayuda a detectar variables categóricas, identificadores
# o categorías sospechosas.

print("Valores distintos por variable en fuente 1:")
display(df1.nunique())

print("Valores distintos por variable en fuente 2:")
display(df2.nunique())
```

### <font color=green> 7. REVISAR LOS VALORES DE UNA VARIABLE CATEGÓRICA </font>

```python
# Puede cambiar "sexo" por otra variable categórica de interés.

print("Valores de la variable sexo en fuente 1:")
display(df1["sexo"].value_counts(dropna=False)) #dropna=False para que también se muestren los valores faltantes

print("Valores de la variable sexo en fuente 2:")
display(df2["sexo"].value_counts(dropna=False))
```

### <font color=green> 8. CREAR UNA TABLA RESUMEN SIMPLE PARA CADA FUENTE </font>

```python
resumen_df1 = pd.DataFrame({
    "fuente": "fuente1_estudiantes",
    "variable": df1.columns,
    "tipo_python": df1.dtypes.values,
    "n_faltantes": df1.isnull().sum().values,
    "porcentaje_faltantes": (df1.isnull().mean() * 100).round(2).values,
    "valores_distintos": df1.nunique().values
})

resumen_df2 = pd.DataFrame({
    "fuente": "fuente2_academica",
    "variable": df2.columns,
    "tipo_python": df2.dtypes.values,
    "n_faltantes": df2.isnull().sum().values,
    "porcentaje_faltantes": (df2.isnull().mean() * 100).round(2).values,
    "valores_distintos": df2.nunique().values
})

tabla_resumen_fuentes = pd.concat([resumen_df1, resumen_df2], ignore_index=True)
#Nota: ignore_index=True hace que Pandas cree un nuevo índice consecutivo al unir las tablas.

print("Tabla resumen de las fuentes:")
display(tabla_resumen_fuentes)
```

### <font color=green> 9. IDENTIFICAR COLUMNAS COMUNES ENTRE LAS FUENTES </font>

```python
# Esto ayuda a proponer una llave de integración.

columnas_comunes = set(df1.columns).intersection(set(df2.columns))

print("Columnas comunes entre las fuentes:")
print(columnas_comunes)
```

### <font color=green> 10. INTEGRAR LAS DOS FUENTES </font>

```python
# Usamos id_estudiante como llave de integración.
# how="outer" conserva todos los registros de ambas fuentes.
# Así podemos ver qué estudiantes aparecen en una fuente, en la otra,
# o en ambas.

base_integrada = pd.merge(
    df1,
    df2,
    on="id_estudiante",
    how="outer",
    suffixes=("_f1", "_f2"),
    indicator=True
)

print("Base integrada:")
display(base_integrada)
```

### <font color=green> 11. REVISAR EL RESULTADO DE LA INTEGRACIÓN </font>

```python
# La columna _merge indica de dónde viene cada registro:
# both       = aparece en ambas fuentes
# left_only  = aparece solo en fuente 1
# right_only = aparece solo en fuente 2

print("Resultado de la integración:")
display(base_integrada["_merge"].value_counts())

print("Valores faltantes en la base integrada:")
display(base_integrada.isnull().sum())

print("Resumen de la base integrada:")
display(base_integrada.describe(include="all"))
```

### <font color=green> 12. PLANTILLA SIMPLE DE DICCIONARIO DE DATOS </font>

```python
# Completar interpretando sus variables.

diccionario_datos = pd.DataFrame({
    "variable": tabla_resumen_fuentes["variable"],
    "fuente": tabla_resumen_fuentes["fuente"],
    "descripcion": "",
    "tipo_general": "",       # Categórica o numérica
    "escala": "",             # Nominal, ordinal, intervalo o razón
    "rol_o_subtipo": "",      # Identificador, binaria, discreta, continua, fecha, texto, etc.
    "observaciones": ""
})

print("Plantilla de diccionario de datos:")
display(diccionario_datos)
```

### <font color=green> 13. EXPORTAR RESULTADOS </font>

```python
# Esto genera un archivo Excel con:
# - tabla resumen de fuentes
# - base integrada
# - plantilla de diccionario de datos

with pd.ExcelWriter("resultado_laboratorio_integracion.xlsx") as writer:
    tabla_resumen_fuentes.to_excel(writer, sheet_name="resumen_fuentes", index=False)
    base_integrada.to_excel(writer, sheet_name="base_integrada", index=False)
    diccionario_datos.to_excel(writer, sheet_name="diccionario_datos", index=False)

print("Archivo generado: resultado_laboratorio_integracion.xlsx")
```
